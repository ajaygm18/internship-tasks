```python
class BankAccount:
    name = "State Bank of India"
    def __init__(self, amount: int):
        self.balance = amount
        self.name = "SBI"

    def withdraw(self, amount: int):
        if amount > self.balance:
            print("Insufficient Balance")
            return
        self.balance = self.balance - amount
        print("Amount Withdrawn Balance is :", self.balance)

    def deposit(self, amount: int):
        if amount < 0:
            print("Amount should be +ve Integer")
            return

        self.balance += amount
        print("Amount Deposited Balance is :", self.balance)

    @staticmethod
    def name():
        print("State Bank of India")

    @classmethod
    def names(cls):
        print(cls.name)

    @property
    def namex(self):
        return self.name

    @namex.setter
    def namex(self, name):
        self.name = name


b1 = BankAccount(5000)
b1.deposit(1000000)
b1.withdraw(5100)
BankAccount.name()
b1.namex = "RBI"
b1.namex
```
