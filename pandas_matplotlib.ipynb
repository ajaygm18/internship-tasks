```python
yearly = df.groupby('Year')['MSRP'].sum().sort_values(ascending=True)
yearly.head()
```

```python
np.linspace(yearly.min(), yearly.max(), num=len(yearly.index), dtype='int64')
```

```python
plt.figure(figsize=(10,10))
plt.plot(yearly.index, yearly.values, marker='o')
plt.xlabel('Years')
plt.ylabel('Revenue')
plt.title("Year Wise Sales")
plt.xticks(yearly.index, rotation=45)
plt.yticks(np.linspace(yearly.min(), yearly.max(), num=len(yearly.index), dtype=int))
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,8))
plt.plot(yearly.index, yearly.values, marker='o')
plt.style.use('ggplot')
plt.grid(True)
for i, value in enumerate(yearly.index):
    plt.text(yearly.index[i], yearly.values[i], f"{yearly.values[i]}", ha='right', fontsize=8, rotation=90)
plt.xticks(yearly.index, rotation=45)
plt.show()
```

```python


engine_revenue = df.groupby('Engine Fuel Type')['MSRP'].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(10, 6))
plt.bar(engine_revenue.index, engine_revenue.values, color='orange')

plt.title("Top 10 Engine Fuel Type")
plt.xlabel("Engine type")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.grid(axis='y')

plt.show()
```

```python


engine_revenue = df.groupby('Engine Fuel Type')['MSRP'].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(10, 6))
plt.barh(engine_revenue.index, engine_revenue.values, color='orange')

plt.title("Top 10 Engine Fuel Type")
plt.xlabel("Total Revenue")
plt.ylabel("Engine type")
plt.xticks(rotation=45)
plt.grid(axis='y')

plt.show()
```

```python
import seaborn as sns
```

```python
yearly_rev = df.groupby('Year')['MSRP'].sum().sort_index()

plt.figure(figsize=(12, 5))
sns.lineplot(x=yearly_rev.index, y=yearly_rev.values, marker='o')
plt.title("Yearly Revenue Trend")
plt.xticks(rotation=45)
plt.ylim(0, yearly_rev.max() * 1.1)
plt.ylabel("Total Revenue")
plt.xlabel("Year")
plt.tight_layout()
plt.show()
```

```python
styles = df['Vehicle Style']
styles
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=styles, order=styles.value_counts().index )
plt.xlabel("Style")
plt.xticks(rotation=45)
plt.ylabel("Number of Vehicles")
plt.title("No of vehicles by style")
plt.show()
```

```python

top_styles = df['Vehicle Style'].value_counts().head(5).index.tolist()

df_top = df[df['Vehicle Style'].isin(top_styles)].copy()

grouped = df_top.groupby(['Vehicle Style', 'Year'])['MSRP'].sum().reset_index()
grouped
```

```python
grouped = grouped.sort_values('Year')
grouped
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=grouped, x='Year', y='MSRP', hue='Vehicle Style')
plt.title('Yearly Revenue by Style (Grouped)')
plt.xlabel("Style")
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
