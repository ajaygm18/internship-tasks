```python
import threading, time
```

```python
def worker(stuff):
    print("Start")
    print(stuff)
    time.sleep(2)
    print("end")

```

```python
thread1 = threading.Thread(target=worker, args=('xd',))
```

```python
thread2 = threading.Thread(target=worker, args=('xxd',))
```

```python
thread1.start()
thread2.start()
thread1.join()
thread2.join()
print("Done")
```
