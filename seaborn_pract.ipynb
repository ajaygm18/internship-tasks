```python
data = sns.load_dataset('tips')

sns.boxplot(x='day', y='tip', data=data)
plt.title('Box Plot: Distribution of Tips by Day')
plt.show()
```

```python
data = sns.load_dataset('flights')

pivot_data = data.pivot_table(values='passengers', index='month', columns='year')
plt.figure(figsize=(10, 6))
sns.heatmap(pivot_data, annot=True, cmap='YlGnBu')
plt.title('Heatmap: Passengers per Month and Year')
plt.show()
```

```python
data = sns.load_dataset('tips')


sns.violinplot(x='day', y='tip', data=data)
plt.title('Violin Plot: Distribution of Tips by Day')
plt.show()
```

```python
data = sns.load_dataset('tips')

sns.relplot(x='total_bill', y='tip', data=data, kind='line', hue='sex', style='time')

plt.title('Scatter Plot: Total Bill vs Tip')
plt.show()
```
